Detecting outliers in Data Using Python:

For this we use stats from scipy library. 

```python
import numpy as np
import pandas as pd
from scipy import stats

# Suppress the FutureWarning related to downcasting
pd.set_option('future.no_silent_downcasting', True)

# Sample data with an "UNKNOWN" value, related to student test scores
data = {
    'Math': [88, 92, 'UNKNOWN', 85, 78, 90, 95, 'UNKNOWN', 87],
    'Science': [75, 82, 78, 'UNKNOWN', 80, 85, 89, 88, 91]
}
df = pd.DataFrame(data)

# Replace 'UNKNOWN' with NaN
df_replaced = df.replace('UNKNOWN', np.nan)

# Explicitly convert data types
df_replaced['Math'] = df_replaced['Math'].astype(float)
df_replaced['Science'] = df_replaced['Science'].astype(float)

# Detecting outliers using Z-score
# TODO: Calculate Z-scores of 'Math' and 'Science' form df_replaced

z_scores = np.abs(stats.zscore(df_replaced[['Math', 'Science']]))

filtered_df = df_replaced.dropna()

df_no_outliers = filtered_df[(z_scores < 3).all(axis=1)]

print(df_no_outliers)
```

Standardization and Normalization the data

Applying above techniques ensure uniformity & improve algorithm efficiency!

Standardization assumes Mean = 0 and Standard deviation = 1
Normalization rescales the data to a fixed range, usually [0, 1]

Applying above techniques depends on the ML algorithm and the nature of the data


```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

# Example DataFrame
df = pd.DataFrame({
    'Age': [25, 32, 47, 51],
    'Salary': [50000, 60000, 120000, 100000]
})

# TODO: Normalize numerical features with range -1 to 1
# The default featue range is 0 to 1
normalizer = MinMaxScaler(feature_range=(-1,1))
df[['Age', 'Salary']] = normalizer.fit_transform(df[['Age', 'Salary']])

print(df)
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Sample data
data = {
    'Weight': [50, 60, 70, 80],
    'Height': [150, 160, 170, 180]
}
df = pd.DataFrame(data)

# Standardizing numerical features
scaler = StandardScaler()
df[['Weight', 'Height']] = scaler.fit_transform(df[['Weight', 'Height']])

print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Sample data with None values
data = {
    'Temperature': [22, None, 25, 30],
    'Humidity': [55, 60, None, 70]
}
df = pd.DataFrame(data)

# TODO: Handle None values by filling them with the mean of the columns
df['Temperature'] = df['Temperature'].fillna(df['Temperature'].mean())
df['Humidity'] = df['Humidity'].fillna(df['Humidity'].mean())


#print(df)

# TODO: Normalize the 'Temperature' and 'Humidity' columns using MinMaxScaler
normalizer = MinMaxScaler()
df[['Temperature', 'Humidity']] = normalizer.fit_transform(df[['Temperature', 'Humidity']])

# TODO: Print the transformed DataFrame
print(df)
```
