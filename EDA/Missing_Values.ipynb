```python
import pandas as pd

# Sample DataFrame creation with missing values
data = {
    'Employee': ['John', None, 'Emma'],
    'Experience': [5, None, 10],
    'Salary': [70000, 80000, None]
}
df = pd.DataFrame(data)

# Dropping rows with missing values
df_cleaned = df.dropna()
print(df_cleaned)

# Filling missing values with appropriate strategies
# TODO: Fill the missing values in the Experience column with the mean value of that column

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Employee'] = df['Employee'].fillna('Unknown')

print(df)
```
