```python
# Importing the libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Importing the dataset
dataset = pd.read_csv(r'C:\Users\tm0792.STUDENTS.008\OneDrive - UNT System\Academics\Second Sem\DTSC 5505 - Applied ML for Data Scientists\Data_Science-Projects\ml-algos\data\Bank_Predictions.csv')
X = dataset.iloc[:, 3:13].values
y = dataset.iloc[:, 13].values

dataset.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Assuming X[:, 1] and X[:, 2] are categorical
# No need to use LabelEncoder for one-hot encoding
onehotencoder = ColumnTransformer(
    transformers=[('cat', OneHotEncoder(), [1, 2])], 
    remainder='passthrough'
)

# Apply OneHotEncoder to columns 1 and 2
X = onehotencoder.fit_transform(X)

# Optionally drop the first column of each one-hot encoded feature to avoid dummy variable trap
X = X[:, 1:]

print(X.shape)
```

```python
# Splitting the dataset into the Training set and Test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

```python
# import keras library and packages
import keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
from keras.layers import Input

# Initializing the ANN
classifier = Sequential()

# Adding the input layer
classifier.add(Input(shape=(12,)))  # Adjust the number based on your actual input features

# Adding the first hidden layer
classifier.add(Dense(units=6, activation='sigmoid', kernel_initializer='uniform'))

# Adding second hidden layer
classifier.add(Dense(units=6, activation='sigmoid', kernel_initializer='uniform'))

# Adding output layer
classifier.add(Dense(units=1, activation='sigmoid', kernel_initializer='uniform'))

# Compiling the ANN
classifier.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Fitting the ANN to the training set
classifier.fit(X_train, y_train, batch_size=10, epochs=100)

```

```python
# Predicting the Test set results
y_pred = classifier.predict(X_test)
y_pred = (y_pred > 0.5)

# Making the confusion Matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```
